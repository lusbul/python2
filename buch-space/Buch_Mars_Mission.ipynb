## Тема 13 · Python і космос
Обчисліть відстань від станції (0, 0) до марсохода (3, 4). Спробуйте змінити координати.

```python
def rover_distance(x, y):
    return math.sqrt(x*x + y*y)

distance = rover_distance(3, 4)
print("Відстань:", distance)
```

## Тема 14 · Python-криптограф
Додаємо функцію розшифрування навчального шифру Цезаря. Для реального захисту повідомлень цей шифр не підходить.

```python
def caesar_decode(text, shift=1):
    result = ""
    for char in text:
        if "A" <= char <= "Z":
            result += chr((ord(char) - 65 - shift) % 26 + 65)
        else:
            result += char
    return result

message = caesar_decode("NBST", 1)
print("Розшифровано:", message)
```

## Тема 15 · Pillow
Додаємо опрацювання знімка. Перевірте, що зміниться, якщо замість 1.6 вказати 0.5 або 2.0.

```python
image = Image.open("mars-rover.png")
image = ImageEnhance.Contrast(image).enhance(1.6)
image.save("mars-enhanced.png")
display(image)
```

## Тема 16 · CSV і JSON
Додаємо читання телеметрії та створення пакета команд для марсохода.

```python
with open("telemetry.csv", encoding="utf-8", newline="") as f:
    records = list(csv.DictReader(f))
low = [r["station"] for r in records if int(r["battery"]) < 50]
command = {"mission": "BUCH-MARS", "action": "check", "stations": low, "message": message, "distance": distance}
with open("mission-result.json", "w", encoding="utf-8") as f:
    json.dump(command, f, ensure_ascii=False, indent=2)
print(json.dumps(command, ensure_ascii=False, indent=2))
```
